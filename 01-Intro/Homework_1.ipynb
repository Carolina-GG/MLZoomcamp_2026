Q7. Sum of weights

```python
df_asia = df[df['origin'] == 'Asia'] # Only Asian cars

X = df_asia[['vehicle_weight', 'model_year']].head(7).to_numpy()
# Selected variables, first 7 observations, creating a np array

XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv.dot(X.T).dot(y)

print(w.sum()) # Sum of all the elements of the result
```
